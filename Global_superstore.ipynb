```python
import pandas as pd

df=pd.read_csv("/content/drive/MyDrive/ML/Churn_Modelling.csv")
# 1. Dataset size and structure
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

# 2. Missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Duplicate checks
print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:",
      df["CustomerId"].duplicated().sum())

# 4. Categorical consistency
for col in ["Gender", "Geography", "Married",
            "Is_spouse_earning", "Exited"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

# 5. Numerical summaries
print("\nNumerical summary:")
print(df[["Balance", "Tenure", "NumOfProducts",
          "EstimatedSalary", "Children"]].describe())

# 6. Inspect extreme balances
print("\nHighest balances:")
print(
    df.nlargest(5, "Balance")[
        ["CustomerId", "Balance", "Exited"]
    ]
)
```

```python
df_clean = df.copy()

# 1. Handle missing categorical values
df_clean["Gender"] = df_clean["Gender"].fillna("Unknown")
df_clean["Married"] = df_clean["Married"].fillna("Unknown")

# 2. Standardise categorical labels
for col in ["Gender", "Geography", "Married", "Is_spouse_earning"]:
    df_clean[col] = df_clean[col].astype("string").str.strip()

df_clean["Married"] = df_clean["Married"].str.title()
df_clean["Is_spouse_earning"] = (
    df_clean["Is_spouse_earning"].str.title()
)

# 3. Check that cleaning worked
print("Dataset shape:", df_clean.shape)
print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nMarital status values:")
print(df_clean["Married"].value_counts(dropna=False))

# 4. Export a cleaned copy
df_clean.to_csv("Churn_Modelling_Cleaned.csv", index=False)

print("\nCleaned dataset exported successfully.")
```
